Distribution des classes

```python

plt.figure(figsize=(10, 5))
train['label'].value_counts().sort_index().plot(kind='bar')
plt.title("Distribution des classes dans le dataset (Sign Language MNIST)")
plt.xlabel("Labels")
plt.ylabel("Nombre d'images")
plt.show()
```

Visualiser quelques images du dataset

```python
# === Extraire quelques images ===
images = train.iloc[:9, 1:].values
labels = train.iloc[:9, 0].values

images = images.reshape(-1, 28, 28)

plt.figure(figsize=(8, 8))

for i in range(9):
    plt.subplot(3, 3, i+1)
    plt.imshow(images[i], cmap='gray')
    plt.title(f"Label : {labels[i]}")
    plt.axis('off')

plt.suptitle("Exemples d'images du dataset Sign Language MNIST")
plt.show()
```
